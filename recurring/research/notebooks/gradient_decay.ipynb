# Recurrent signal decay: a teaching model

This notebook reproduces the scalar illustration used in the site. It is not a replication of Elman (1990), Bengio et al. (1994), or an empirical neural-network experiment. The scalar recurrence is `s_t = r s_(t-1)`, so the derivative after n steps is `r^n`. Real RNN gradients use products of state-dependent Jacobians.

```python
import torch
import matplotlib.pyplot as plt

steps = torch.arange(0, 65, dtype=torch.float64)
for r in (0.72, 0.88, 0.96):
    plt.plot(steps, torch.pow(torch.tensor(r, dtype=torch.float64), steps), label=f'r = {r}')
plt.xlabel('recurrent steps'); plt.ylabel('signal magnitude'); plt.legend(); plt.grid(alpha=.2)
```

The curves show repeated multiplication in a deliberately minimal system. They explain a mechanism that can contribute to vanishing signals; they do not claim that every recurrent network behaves this way at every step.
