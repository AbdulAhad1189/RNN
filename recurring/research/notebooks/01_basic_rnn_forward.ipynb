# Basic RNN forward pass

A modern teaching example, not a reproduction of Elman's exact network.

Requires Python 3 and PyTorch.

```python
import torch
print('PyTorch',torch.__version__)
```

```python
import torch
torch.manual_seed(7)
Wx=torch.tensor([[.8,-.2],[.1,.6]]); Wh=torch.tensor([[.3,.1],[-.4,.2]]); h=torch.zeros(2)
for t,x in enumerate([torch.tensor([1.,0.]),torch.tensor([0.,1.]),torch.tensor([1.,1.])],1):
 h=torch.tanh(Wx@x+Wh@h); print(t,h.tolist())
```
