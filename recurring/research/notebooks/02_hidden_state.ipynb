# Hidden-state visualization

Plot activations from a tiny fixed RNN; values depend on illustrative weights.

Requires Python 3 and PyTorch.

```python
import torch
print('PyTorch',torch.__version__)
```

```python
import torch
import matplotlib.pyplot as plt
Wx=torch.tensor([.8,-.5,.3]); Wh=torch.tensor([[.2,.1,0.],[0.,.3,.1],[.1,0.,.25]])
h=torch.zeros(3); states=[]
for x in [0.,1.,1.,0.,1.,0.]:
 h=torch.tanh(Wx*x+Wh@h); states.append(h.tolist())
plt.imshow(states,aspect='auto',cmap='cividis'); plt.xlabel('hidden unit'); plt.ylabel('time step'); plt.colorbar(label='activation')
```
