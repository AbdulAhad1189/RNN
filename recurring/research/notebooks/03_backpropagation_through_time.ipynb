# Backpropagation through time

Autograd traces a final loss through the unrolled recurrence.

Requires Python 3 and PyTorch.

```python
import torch
print('PyTorch',torch.__version__)
```

```python
import torch
wx=torch.tensor(.7); wh=torch.tensor(.8,requires_grad=True); h=torch.tensor(0.)
for x in [.2,-.1,.4,.3]: h=torch.tanh(wx*x+wh*h)
loss=(h-.5).square(); loss.backward()
print('final state:',h.item()); print('d(loss)/d(recurrent weight):',wh.grad.item())
```
