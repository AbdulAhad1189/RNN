# Simple sequence prediction

Train a toy Elman-style RNN to predict the next bit in a repeating pattern. Outcomes may vary by initialization.

Requires Python 3 and PyTorch.

```python
import torch
print('PyTorch',torch.__version__)
```

```python
import torch
from torch import nn
torch.manual_seed(3)
class TinyRNN(nn.Module):
 def __init__(self):
  super().__init__(); self.cell=nn.RNNCell(1,8); self.readout=nn.Linear(8,1)
 def forward(self,x):
  h=torch.zeros(1,8); out=[]
  for item in x:
   h=self.cell(item.reshape(1,1),h); out.append(self.readout(h))
  return torch.cat(out)
pattern=torch.tensor([0.,1.,0.,1.,0.,1.,0.,1.]); x=pattern[:-1]; target=pattern[1:]
model=TinyRNN(); opt=torch.optim.Adam(model.parameters(),lr=.03)
for epoch in range(250):
 opt.zero_grad(); loss=nn.functional.binary_cross_entropy_with_logits(model(x),target); loss.backward(); opt.step()
print('final toy loss:',round(loss.item(),4))
```
