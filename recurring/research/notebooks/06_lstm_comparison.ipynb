# LSTM comparison

LSTM is later work (Hochreiter & Schmidhuber, 1997), not part of Elman’s 1990 paper.

Requires Python 3 and PyTorch.

```python
import torch
print('PyTorch',torch.__version__)
```

```python
import torch
from torch import nn
torch.manual_seed(5)
rnn=nn.RNN(2,4,batch_first=True); lstm=nn.LSTM(2,4,batch_first=True); x=torch.randn(1,6,2)
rnn_out,rnn_h=rnn(x); lstm_out,(lstm_h,lstm_c)=lstm(x)
print('RNN output/state:',tuple(rnn_out.shape),tuple(rnn_h.shape))
print('LSTM output/hidden/cell:',tuple(lstm_out.shape),tuple(lstm_h.shape),tuple(lstm_c.shape))
```
