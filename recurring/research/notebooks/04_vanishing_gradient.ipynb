# Vanishing-gradient experiment

A controlled scalar tanh recurrence. Derivatives are generated by PyTorch autograd, not reported results from Elman (1990).

Requires Python 3 and PyTorch.

```python
import torch
print('PyTorch',torch.__version__)
```

```python
import torch
for weight in [.7,.95,1.05]:
 for length in [1,2,4,8,16,32]:
  h0=torch.tensor(.1,requires_grad=True); h=h0
  for _ in range(length): h=torch.tanh(weight*h+.1)
  (g,)=torch.autograd.grad(h,h0)
  print(f'weight={weight:.2f}, steps={length:2d}, |dhT/dh0|={abs(g.item()):.6g}')
```
